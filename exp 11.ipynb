```python
#generate a positive listof numbers from a given list
numbers=list(map(int,input("Enter the numbers:").split()))
print("the numbers:",numbers)
result=[i for i in numbers if i>0]
print("positive numbers are:",result)
```

```python
#squares of N numbers
n=int(input("enter the N:"))
print("the limit is:",n)
result=[i*i for i in range(1,n+1)]
print("squares:",result)
```

```python
#from a list of vowels selected from a given word
word=input("enter a word:")
print("the word is:",word)
result=[i for i in word if i.lower()in "aeiou"]
print("vowels:",result)
```

```python
#list ordinal values of each element of a word using ord()
words=input("enter the word:")
print("the word is :",words)
result=[ord(i) for i in words]
print("ordinal values:",result)
```
