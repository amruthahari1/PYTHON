```python
#color from colorset1 not contained in colorset2
colorset1={"red","orange","blue","green"}
colorset2={"green","yellow","black"}
c=colorset1-colorset2
print("colors in set 1 not in set 2:",c)
```

```python
#color from both set1 and set2
colorset1={"red","orange","blue","green","black"}
colorset2={"green","yellow","black"}
c=colorset1&colorset2
print("colors in both sets:",c)
```
