```python
#sort the dictionary in ascending and descending order of their roll no roll=input("enter roll number:").split()
roll=input("enter the roll number:").split()
name=input("enter student name:").split()
my_dict=dict(zip(roll, name))
print ("original dictionary is: ",my_dict)
print ("ascendsing:",dict(sorted(my_dict.items ())))
print ("descending:" ,dict(sorted(my_dict.items())))
```
