```python
#sort the dictionary in ascending and descending order of their roll no
roll=input("enter roll number:").split()
name=input("enter student name:").split()
my_dict=dict(zip(roll,name))
print("original dictionary is: ",my_dict)
print("ascendsing:",dict(sorted(my_dict.items())))
print("descending:",dict(sorted(my_dict.items())))
```

```python
#merge 2 dict and display
dic_1=dict(zip(input("enter 3 roll number:").split(),input("enter 3 name:").split()))
dic_2=dict(zip(input("enter 2 roll number:").split(),input("enter 2 name:").split()))
print("dictionary 1=",dic_1)
print("dictionary 2=",dic_2)
dic_1.update(dic_2)
print("Merge dict:",dic_1)
```
