```python
#count each word in a text
text=input("enter the text:")
words=text.split()
print("text is:",text)
for w in set(words):
    print(w,":",words.count(w))
```

```python
#replace the value greater then 100 with"over"
numbers=list(map(int,input("enter the numbers:").split()))
print("numbers are:",numbers)
for i in range(len(numbers)):
    if numbers[i]>100:
        numbers[i]="over"
print("result is:",numbers)
```

```python
#count occurrences of 'a' within the list
names=input("Enter the names(separated by space):").split()
print("The names are:",names)
count=0
for n in names:
    count+=n.lower().count("a")
print("Occurrence of 'a':",count)
```

```python
#Find the gcd of two numbers.
a=int(input("Enter the first number:"))
b=int(input("Enter the second number:"))
print("First number:",a)
print("Second number:",b)
x,y=a,b
while b!=0:
    a,b=b,a%b
print("Gcd(",x,",",y,"):",a)
```

```python
#Remove even numbers in a list.
numbers=list(map(int,input("Enter the numbers (separated by space):").split()))
result=[]
print("The numbers are:",numbers)
for i in numbers:
    if i%2!=0:
        result.append(i)
print("List after removing even numbers:",result)
```

```python
#Find the factorial of a number.
number=int(input("Enter the number:"))
fact=1
print("The number is:",number)
for i in range(1,number+1):
    fact=fact*i
print("Factorial of",number,"=",fact)
```

```python
#Generate fibonacci series of n terms.
n=int(input("Enter the number of terms:"))
a=0
b=1
print("The number of term is:",n)
print("The fibonacci series is:",end="")
for i in range(n):
    print(a,end="\t")
    a,b=b,a+b
```

```python
#Find the sum of list.
numbers=list(map(int,input("Enter the numbers (separated by space):").split()))
total=0
print("The numders are:",numbers)
for i in numbers:
    total=total+i
print("Sum of all items:",total)
```

```python
#Find four-digit perfect squares with all digits even
import math
start=int(input("Enter the starting range:"))
end=int(input("Enter the ending range:"))
print("starting range:",start,"\nEnding range:",end)
print("Result is:",end=" ")
for n in range(start,end+1):
    if int(math.sqrt(n))**2==n:
        if all(d in "02468" for d in str(n)):
            print(n,end=" ")
```

```python
#Count character frequency in a string.
s=input("Enter a string:")
print("String is:",s)
print("Character frequency:")
chars=[]
for ch in s:
    if ch not in chars:
        chars.append(ch)
for ch in chars:
    print(ch, ":", s.count(ch))
```

```python
#Compare two lists by length,sum and common values.
list1=list(map(int,input("Enter first list:").split()))
list2=list(map(int,input("Enter second list:").split()))
print("First list:",list1)
print("Second list:",list2)
if len(list1)==len(list2):
    print("Both list have the same length.")
else:
    print("Both list do not have the same length.")
if sum(list1)==sum(list2):
    print("Both list have the same sum.")
else:
    print("Both list do not have the same sum.")
if set(list1)&set(list2):
    print("Both list have a common value.")
else:
    print("Both list do not have a common value.")
```

```python
#Generate a number pyramid for n rows.
num=int(input("Enter the number:"))
print("The number:",num)
for i in range(1,num+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print()
```
