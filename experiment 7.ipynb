```python
#color from colorset1 not contained in colorset2
colorset1=set(input("enter color in set 1:").split())
colorset2=set(input("enter color in set 2:").split())
print("set 1=",colorset1)
print("set 2=",colorset2)
print("the colors in set 1 but not in set 2:",colorset1.difference(colorset2))
```

```python
#color from both set1 and set2
colorset1=(input("enter color in set 1:").split())
colorset2=(input("enter color in set 2:").split())
print("set 1=",colorset1)
print("set 2=",colorset2)
print("the colors in both sets are:",colorset1.union(colorset2))
```
